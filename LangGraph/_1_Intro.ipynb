### **Instalation**

```python
from dotenv import load_dotenv
load_dotenv()
from langchain_groq import ChatGroq
from langchain.tools import tool

llm = ChatGroq(model= "meta-llama/llama-4-scout-17b-16e-instruct", temperature=0.5)

@tool
def get_weather(location: str) -> str:
    """
    This is the tool for determining the weather related query for an given location use this when user asks for weather updates
    """
    return f"{location} is Sunny"

llm_with_tools = llm.bind_tools([get_weather])
llm_with_tools.invoke("Give me the weather of Kolkata")
```
