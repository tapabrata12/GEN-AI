### **Linear Workflow**

```python
from dotenv import load_dotenv
load_dotenv()

from langchain_groq import ChatGroq
from langgraph.graph import StateGraph,START,END
from typing import TypedDict
import os

if os.getenv("GROQ_API_KEY") is None:
    raise ValueError("GROQ_API_KEY environment variable not set")


llm = ChatGroq(model= "meta-llama/llama-4-scout-17b-16e-instruct", temperature=0.5)
```

```python
class LLMState(TypedDict):
    question: str
    answer: str


def giveAnswer(state: LLMState) -> LLMState:
    question = state['question']
    answer = llm.invoke(question).content
    state['answer'] = answer
    return state
```

```python
graph = StateGraph(LLMState)

graph.add_node('giveAnswer', giveAnswer)
graph.add_edge(START, 'giveAnswer')
graph.add_edge('giveAnswer', END)

workflow = graph.compile()
```

```python
data = {'question': "What is the capital of France?"}
workflow.invoke(data)
```
